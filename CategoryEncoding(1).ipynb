# Category Encoding
two methods:
1. One-hot encoding
2. Target (Mean) Encoding

## Data and Task

### LarrisaRent.csv
Rent for one-bedroom apartments in Larissa

#### Dataset Structure

| Column Name | Type | Description |
|-------------------|-------------|---|
| `Rent` | Numeric | **Target variable**: Monthly rental price in Euro (€) |
| `Area` | Numeric | Total interior living space in square meters |
| `Parking` | Binary | `Yes` if the apartment has parking, `No` if not |
| `District` | Categorical | Neighborhood where the appartment is located (5 categories listed below)|

#### Categorical Variable: `Neighborhood`

The `Neighborhood` column includes five meaningful categories representing different areas in a mid-sized city:
- `Centre`
- `North`
- `South`
- `Historic`
- `Pineios`

### Task: Predict Rent Prices
Predict Rent Prices in Larissa for one-bedroom apartments

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv('LarissaRent.csv')
df.head()
```

```python
# Map 'Yes' to 1, 'No' to 0
df['Park'] = df['Parking'].map({'Yes': 1, 'No': 0})
df.head()
```

```python
# Get unique categories
districts = df['District'].unique()
districts
```

```python
# Create one-hot columns manually
for district in districts:
    df[f'District_{district}'] = (df['District'] == district).astype(int)

df.head()
```

```python
# Drop original categorical columns
df_encoded = df.drop(columns=['Parking', 'District'])
df_encoded.head()
```

```python
# Get input features and target
X = df_encoded.drop(columns=['Rent'])
y = df_encoded['Rent'] 

# Initialize and train model
model = LinearRegression()
model.fit(X, y)

# Predict rent
y_pred = model.predict(X)

# Step 4: Evaluate
print(f"R2 statistic: {r2_score(y, y_pred):.2f}")
```

```python
df = pd.read_csv('LarissaRent.csv')
df.head()
```

```python
# Encode 'Parking' using mean rent per category
parking_means = df.groupby('Parking')['Rent'].mean()
df['Parking_encoded'] = df['Parking'].map(parking_means)
df.head()
```

```python
# Encode 'District' using mean rent per category
district_means = df.groupby('District')['Rent'].mean()
df['District_encoded'] = df['District'].map(district_means)
df.head()
```

```python
# Drop original categorical columns
df_encoded = df.drop(columns=['Parking', 'District'])
df_encoded.head()
```

```python
# Get input features and target
X = df_encoded.drop(columns=['Rent'])
y = df_encoded['Rent'] 

# Initialize and train model
model = LinearRegression()
model.fit(X, y)

# Predict rent
y_pred = model.predict(X)

# Step 4: Evaluate
print(f"R2 Score: {r2_score(y, y_pred):.2f}")
```
